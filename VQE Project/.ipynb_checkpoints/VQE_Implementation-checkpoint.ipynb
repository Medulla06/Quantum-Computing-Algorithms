## Contructing Hamiltonian

```python
def get_qubit_counts(n: int) -> tuple[int, int, int]:
    """
    Compute Np, Nq, and total qubit count N for a given n.
    As per the paper: p <= sqrt(n), q < n/2, both odd => drop LSB.
    """
    B = ceil(log2(n + 1))
    Np = ceil(log2(n**0.5 + 1))   # bits for p (incl. p0=1)
    Nq = B - 1                     # bits for q (incl. q0=1)
    N = (Np - 1) + (Nq - 1)       # free qubits (strip fixed LSBs)
    return Np, Nq, N


def bitstring_to_pq(bits: str, Np: int, Nq: int) -> tuple[int, int]:
    """
    Convert a measurement bitstring to factor integers p, q.
    Qiskit bit order: rightmost bit = qubit 0.
    Layout: |p_{Np-1} ... p_1  q_{Nq-1} ... q_1⟩
    LSBs p0=q0=1 are appended.
    """
    bits = bits[::-1]              # reverse to LSB-first order
    q_bits = [int(b) for b in bits[:Nq - 1]]    # q1 .. q_{Nq-1}
    p_bits = [int(b) for b in bits[Nq - 1:]]    # p1 .. p_{Np-1}

    p = 1 + sum(p_bits[j] * (2 ** (j + 1)) for j in range(len(p_bits)))
    q = 1 + sum(q_bits[k] * (2 ** (k + 1)) for k in range(len(q_bits)))
    return p, q


def cost_fn(p: int, q: int, n: int, cost_type: str = "hamiltonian") -> float:
    """
    Evaluate cost function for a given (p, q) candidate.
    Three types as described in Section VI.C of the paper:
      - 'hamiltonian': (n - pq)^2
      - 'logarithm':   log(|n - pq| + 1)
      - 'inverse':     -1 / (|n - pq| + 0.001)
    """
    diff = abs(n - p * q)
    if cost_type == "hamiltonian":
        return float(diff ** 2)
    elif cost_type == "logarithm":
        return float(np.log(diff + 1))
    elif cost_type == "inverse":
        return -1.0 / (diff + 0.001)
    else:
        raise ValueError(f"Unknown cost_type: {cost_type}")

```

## Quantum circuit

```python
def build_linear_cnot_ansatz(N: int, L: int, params: np.ndarray) -> "QuantumCircuit":
    """
    Build the hardware-efficient linear-CNOT ansatz described in Section III / Fig. 2.
    N qubits, L layers.
    Each layer: RY(theta) on each qubit, then CNOT ladder (qubit i -> i+1).
    Total parameters: N * L RY gates.
    """
    qc = QuantumCircuit(N)

    # Hadamard initialisation (puts qubits in |+> superposition)
    qc.h(range(N))

    param_idx = 0
    for layer in range(L):
        # RY rotations
        for q in range(N):
            qc.ry(params[param_idx], q)
            param_idx += 1
        # Linear CNOT entangling layer (skip for last layer per Fig. 2)
        if layer < L - 1:
            for q in range(N - 1):
                qc.cx(q, q + 1)

    qc.measure_all()
    return qc


def num_params(N: int, L: int) -> int:
    return N * L
```

## CVaR cost function evaluation

```python
def cvar(eigenvalues: list[float], alpha: float) -> float:
    """
    Conditional Value at Risk (CVaR) as in Eq. (8) of the paper.
    Takes the expectation of the lowest alpha-fraction of measured eigenvalues.
    """
    sorted_evals = sorted(eigenvalues)
    k = max(1, int(np.ceil(alpha * len(sorted_evals))))
    return float(np.mean(sorted_evals[:k]))


def evaluate_cost(
    params: np.ndarray,
    n: int,
    Np: int,
    Nq: int,
    N: int,
    L: int,
    shots: int,
    alpha: float,
    cost_type: str,
    simulator: "AerSimulator",
    best_found: dict,
) -> float:
    """
    Build circuit, run on simulator, compute CVaR cost.
    Also tracks the best (p, q) solution found so far.
    """
    qc = build_linear_cnot_ansatz(N, L, params)
    job = simulator.run(qc, shots=shots)
    counts = job.result().get_counts()

    eigenvalues = []
    for bitstring, count in counts.items():
        p, q = bitstring_to_pq(bitstring, Np, Nq)
        ev = cost_fn(p, q, n, cost_type)
        eigenvalues.extend([ev] * count)

        # Track best solution
        if p * q == n and p > 1 and q > 1:
            best_found["p"] = p
            best_found["q"] = q
            best_found["found"] = True

    return cvar(eigenvalues, alpha), best_found
```

## Optimisation

```python
def nft_minimize(obj, params0, max_iter=80):
    """
    Proxy for the NFT (Sequential Minimal Optimization) optimizer
    [Nakanishi et al., PRR 2020]. Coordinate-wise sinusoidal updates.
    """
    params = params0.copy()
    for it in range(min(max_iter, 120)):
        i = it % len(params)
        bv = obj(params)
        for shift in [np.pi / 2, np.pi, 3 * np.pi / 2]:
            trial = params.copy(); trial[i] += shift
            v = obj(trial)
            if v < bv: bv = v; params = trial
    return params
```

```python
def run_vqe(
    n: int,
    L: int = 3,
    shots: int = 1000,
    alpha: float = 0.1,
    cost_type: str = "hamiltonian",
    max_iter: Optional[int] = None,
    optimizer: str = "COBYLA",
    initial_params: Optional[np.ndarray] = None,
    verbose: bool = True,
) -> dict:
    """
    Run one VQE trial to factor n.

    Parameters
    ----------
    n          : integer to factor
    L          : number of ansatz layers
    shots      : measurement shots per evaluation
    alpha      : CVaR percentile (0 < alpha <= 1)
    cost_type  : 'hamiltonian', 'logarithm', or 'inverse'
    max_iter   : COBYLA max iterations (default: 50*N*L per paper)
    initial_params : optional starting angles; random if None
    verbose    : print progress

    Returns
    -------
    dict with keys: found, p, q, n, iterations, success_fidelity, history
    """

    assert n % 2 == 1 and n > 1, "n must be an odd integer > 1"

    Np, Nq, N = get_qubit_counts(n)
    n_params = num_params(N, L)
    if max_iter is None:
        max_iter = 50 * N * L

    simulator = AerSimulator()
    best_found = {"found": False, "p": None, "q": None}
    iteration_count = [0]
    history = [] # Initialize history list

    if initial_params is None:
        params0 = np.random.uniform(-np.pi, np.pi, n_params)
    else:
        params0 = initial_params.copy()

    if verbose:
        print(f"\n{'─'*55}")
        print(f"  Factoring n = {n}")
        print(f"  Qubits N={N}  |  Layers L={L}  |  alpha={alpha}")
        print(f"  Cost: {cost_type}  |  Shots: {shots}  |  MaxIter: {max_iter}")
        print(f"{'─'*55}")

    def objective(params):
        iteration_count[0] += 1
        cv, _ = evaluate_cost(params, n, Np, Nq, N, L, shots, alpha, cost_type, simulator, best_found)
        history.append(cv) # Store the CVaR cost
        if verbose and iteration_count[0] % 10 == 0:
            status = f"  -> Found p={best_found['p']}, q={best_found['q']}" if best_found["found"] else ""
            print(f"  Iter {iteration_count[0]:4d}  |  CVaR = {cv:.4f}{status}")
        return cv

    # result = minimize(
    #     objective,
    #     params0,
    #     method="COBYLA",
    #     options={"maxiter": max_iter, "rhobeg": 0.5},
    # )
    if optimizer == "COBYLA":
        optimizer = COBYLA(maxiter=max_iter)
    elif optimizer == "NFT":
        # NFT params: maxiter (outer loops), reset_interval (default 32)
        # Larger reset_interval = more thorough but slower per restart
        optimizer = NFT(maxiter=max_iter, maxfev=max_iter * N * L, reset_interval=32)
    else:
        raise ValueError(f"Unknown optimizer: {optimizer}. Use 'COBYLA' or 'NFT'.")

    result = optimizer.minimize(fun=objective, x0=params0)


    # Final measurement pass to extract solution if not yet found
    if not best_found["found"]:
        qc = build_linear_cnot_ansatz(N, L, result.x)
        job = simulator.run(qc, shots=shots * 5)
        counts = job.result().get_counts()
        for bitstring in sorted(counts, key=counts.get, reverse=True):
            p, q = bitstring_to_pq(bitstring, Np, Nq)
            if p * q == n and p > 1 and q > 1:
                best_found.update({"found": True, "p": p, "q": q})
                break

    if verbose:
        if best_found["found"]:
            print(f"\n  -> SUCCESS: {n} = {best_found['p']} × {best_found['q']}")
        else:
            print(f"\n  X Not found after {iteration_count[0]} iterations.")
        print(f"{'─'*55}\n")

    return {
        "found": best_found["found"],
        "p": best_found["p"],
        "q": best_found["q"],
        "n": n,
        "iterations": iteration_count[0],
        "history": history, # Return the history list
        "params": result.x
    }
```

```python
def benchmark(
    n: int,
    num_runs: int = 10,
    L: int = 3,
    shots: int = 1000,
    alpha: float = 0.1,
    cost_type: str = "hamiltonian",
    verbose: bool = False,
) -> dict:
    """
    Run VQE num_runs times with random initialisations and report success rate.
    Mirrors the paper's evaluation methodology (Section IV).
    """
    successes = 0
    iters_list = []

    print(f"\nBenchmarking n={n} over {num_runs} runs (L={L}, alpha={alpha}, cost={cost_type})")
    for i in range(num_runs):
        result = run_vqe(n, L=L, shots=shots, alpha=alpha, cost_type=cost_type, verbose=verbose)
        if result["found"]:
            successes += 1
            iters_list.append(result["iterations"])
        print(f"  Run {i+1:2d}/{num_runs}  |  {'->' if result['found'] else '✗'}  "
              f"|  iters={result['iterations']}")

    success_rate = successes / num_runs
    avg_iters = np.mean(iters_list) if iters_list else float("nan")
    print(f"\n  Success rate: {success_rate:.0%}  |  Avg iters (successes): {avg_iters:.1f}\n")
    return {"n": n, "success_rate": success_rate, "avg_iters": avg_iters, "runs": num_runs}
```

```python
def factorize(n: int, **kwargs) -> Optional[tuple[int, int]]:
    """
    Attempt to factor n using VQE.  Returns (p, q) or None.
    Extra keyword args are forwarded to run_vqe().
    """
    result = run_vqe(n, **kwargs)
    if result["found"]:
        return result["p"], result["q"]
    return None
```

```python
# Quantum circuit diagram (linear-CNOT ansatz)
n=15
L=3
Np, Nq, N = get_qubit_counts(n)
params = np.zeros(N * L)

def build_ansatz(N, L, params):
    """
    Hardware-efficient linear-CNOT ansatz (Fig. 2 of paper).
    N qubits, L layers. Each layer: H-init → RY rotations → CNOT ladder.
    Total parameters = N * L.
    """
    qc = QuantumCircuit(N)
    qc.h(range(N))
    idx = 0
    for layer in range(L):
        for q in range(N):
            qc.ry(params[idx], q); idx += 1
        if layer < L - 1:
            for q in range(N - 1):
                qc.cx(q, q + 1)
    qc.measure_all()
    return qc

qc     = build_ansatz(N, L, params).remove_final_measurements(inplace=False)
fig    = qc.draw(output="mpl", fold=-1, plot_barriers=False)
fig.suptitle(
    f"Linear-CNOT Ansatz  (n={n},  N={N} qubits,  L={L} layers)",
    fontsize=13, #y=1.02,
)
fig.savefig("circuit_diagram.png", dpi=150, bbox_inches="tight")
plt.show()
display(fig)
print("circuit_diagram.png")
```

```python
# COBYLA vs NFT convergence

n= 511
L= 4
shots=10000
alpha= 0.01
filename = "Optimizer_comparison.png"

Np, Nq, N   = get_qubit_counts(n)
cost_types  = ["hamiltonian", "logarithm", "inverse", "exponential"]
ct_titles   = {
    "hamiltonian": "Hamiltonian (n−pq)²",
    "logarithm"  : "Logarithm log(|n−pq|+1)",
    "inverse"    : "Inverse −1/(|n−pq|+ε)",
}
opt_styles  = {"COBYLA": ("-",  2.0, COLORS[0]),
                "NFT"   : ("--", 1.8, COLORS[1])}

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
for ax, ct in zip(axes, cost_types):
    for oname, (ls, lw, color) in opt_styles.items():
        res = run_vqe(n, optimizer=oname,
                      L=L, shots=shots, alpha=alpha, cost_type=ct,
                      max_iter=150, verbose=False)
        hist = res["history"]
        ax.plot(hist, color=color, ls=ls, lw=lw, alpha=0.9, label=oname)
        if res["found"]:
            # Mark first point below 0 (inverse) or near-zero (others)
            thresh = -900 if ct == "inverse" else 1.0
            for i, v in enumerate(hist):
                if (ct == "inverse" and v < thresh) or (ct != "inverse" and v < thresh):
                    ax.axvline(i, color=color, ls=":", lw=1, alpha=0.6)
                    break
    ax.set_xlabel("Iterations")
    ax.set_ylabel("CVaR Cost")
    ax.set_title(ct_titles[ct])
    ax.set_yscale("symlog")
    ax.legend()
fig.suptitle(
    f"COBYLA vs NFT Convergence (n={n}, α={alpha})", fontsize=13
)
fig.tight_layout()
fig.savefig(filename, dpi=150, bbox_inches="tight")
print(filename)
```

```python
# Energy landscaple plots

n=511
L=4
shots=10000
alpha=0.01
grid=25

Np, Nq, N = get_qubit_counts(n)
t0 = np.linspace(-np.pi, np.pi, grid)
t1 = np.linspace(-np.pi, np.pi, grid)
Z  = np.zeros((grid, grid))
for i, a in enumerate(t0):
    for j, b in enumerate(t1):
        p = np.zeros(N * L); p[0] = a; p[min(1, N * L - 1)] = b
        cv, _ = evaluate_cost(p, n, Np, Nq, N, L, shots, alpha, "hamiltonian",SIM, {"found": False, "p": None, "q": None})
        Z[i, j] = cv

# Plot 1: CVaR Cost Landscape
fig1, ax1 = plt.subplots(figsize=(6, 4.5))
im = ax1.contourf(t0, t1, Z.T, levels=25, cmap="plasma")
ax1.set_xlabel("θ₀ (rad)"); ax1.set_ylabel("θ₁ (rad)")
ax1.set_title(f"CVaR Cost Landscape  (n={n}, α={alpha})")
fig1.colorbar(im, ax=ax1, label="CVaR")
fig1.suptitle("Energy Landscape — Contour Plot", fontsize=13)
fig1.tight_layout()
filename1 = "cost_landscape_contour.png"
fig1.savefig(filename1, dpi=150, bbox_inches="tight")
print(filename1)
#display(fig1)

# Plot 2: 1-D Slices Through Landscape
fig2, ax2 = plt.subplots(figsize=(6, 4.5))
mid = grid // 2
ax2.plot(t0, Z[:, mid], color=COLORS[0], lw=2, label="θ₁ = 0 slice")
ax2.plot(t1, Z[mid, :], color=COLORS[1], lw=2, ls="--", label="θ₀ = 0 slice")
ax2.set_xlabel("Parameter (rad)"); ax2.set_ylabel("CVaR Cost")
ax2.set_title("1-D Slices Through Landscape"); ax2.legend()
fig2.suptitle("Energy Landscape — 1-D Slices", fontsize=13)
fig2.tight_layout()
filename2 = "cost_landscape_slices.png"
fig2.savefig(filename2, dpi=150, bbox_inches="tight")
print(filename2)
#display(fig2)
```

```python
def run_vqe_fidelity_tracked(n, L, shots, alpha, cost_type, optimizer_name, 
                              max_iter, fidelity_threshold=0.01):
    """
    Tracks the iteration where the fidelity (fraction of shots that return
    the correct factorization) first exceeds fidelity_threshold.
    Only counts as a successful run if threshold is crossed.
    Returns (found_at_iter, max_fidelity_seen) or (None, max_fidelity_seen).
    """
    Np, Nq, N = get_qubit_counts(n)
    simulator  = AerSimulator()
    found_at   = [None]
    iter_count = [0]
    max_fid    = [0.0]
    local_best = {"found": False, "p": None, "q": None}

    params0 = np.random.uniform(-np.pi, np.pi, num_params(N, L))

    def objective(params):
        iter_count[0] += 1
        cv, best = evaluate_cost(
            params, n, Np, Nq, N, L, shots, alpha,
            cost_type, simulator, local_best,
        )

        # Rerun the same circuit to get raw counts for fidelity calculation
        # (evaluate_cost already ran it, but doesn't expose counts — so we
        #  compute fidelity from best_found as a proxy: if found this iter,
        #  fidelity >= 1/shots. We need the actual count so re-query.)
        qc = build_linear_cnot_ansatz(N, L, params)
        job = simulator.run(qc, shots=shots)
        counts = job.result().get_counts()

        # Fidelity = fraction of shots that gave correct (p, q)
        correct_shots = sum(
            count for bs, count in counts.items()
            if (lambda pv, qv: pv * qv == n and pv > 1 and qv > 1)(
                *bitstring_to_pq(bs, Np, Nq)
            )
        )
        fidelity = correct_shots / shots
        max_fid[0] = max(max_fid[0], fidelity)

        if fidelity >= fidelity_threshold and found_at[0] is None:
            found_at[0] = iter_count[0]

        return cv

    if optimizer_name == "COBYLA":
        opt = COBYLA(maxiter=max_iter)
    else:
        opt = NFT(maxiter=max_iter, maxfev=max_iter * N * L, reset_interval=32)

    opt.minimize(fun=objective, x0=params0)
    return found_at[0], max_fid[0]


# ── Fig 10 ────────────────────────────────────────────────────────────────────
n        = 253
filename = "iterations_fidelity.png"
runs     = 20
max_iter = 80
fidelity_threshold = 0.01   # 1% of shots — matches paper

configs = [
    ("COBYLA", "hamiltonian", "o", "-",  "COBYLA, Hamiltonian"),
    ("COBYLA", "logarithm",   "s", "--", "COBYLA, Logarithm"),
    ("COBYLA", "inverse",     "^", ":",  "COBYLA, Inverse"),
    ("NFT",    "hamiltonian", "v", "-",  "NFT, Hamiltonian"),
    ("NFT",    "logarithm",   ">", "--", "NFT, Logarithm"),
    ("NFT",    "inverse",     "*", ":",  "NFT, Inverse"),
]
shot_values = [1000, 5000, 10000, 50000, 100000]

fig, ax = plt.subplots(figsize=(8, 5))

for (opt, ct, mk, ls, lbl), color in zip(configs, COLORS):
    mean_iters = []
    for shots in shot_values:
        ilist = []
        for _ in range(runs):
            found_at, max_fid = run_vqe_fidelity_tracked(
                n, L=3, shots=shots, alpha=0.01,
                cost_type=ct, optimizer_name=opt,
                max_iter=max_iter,
                fidelity_threshold=fidelity_threshold,
            )
            if found_at is not None:
                ilist.append(found_at)

        mi = np.mean(ilist) if ilist else np.nan
        mean_iters.append(mi)
        print(f"  [{opt}, {ct}, shots={shots}] "
              f"avg_iters={mi:.1f}  ({len(ilist)}/{runs} crossed threshold)")

    valid = [(s, v) for s, v in zip(shot_values, mean_iters) if not np.isnan(v)]
    if valid:
        xs, ys = zip(*valid)
        ax.semilogx(xs, ys, color=color, ls=ls, marker=mk, ms=6, lw=2, label=lbl)

ax.axhline(max_iter, color="grey", ls=":", lw=1, label="max_iter ceiling")
ax.set_xlabel("Number of Shots")
ax.set_ylabel(f"Avg Iterations to Fidelity ≥ {fidelity_threshold}")
ax.set_title(f"Iterations to Fidelity Threshold  (n={n})")
ax.legend(fontsize=9)
fig.tight_layout()
fig.savefig(filename, dpi=150, bbox_inches="tight")
print(filename)
```

## Hardware Results

```python
# ── Batch hardware evaluator ──────────────────────────────────────────────────
# Submits a list of parameter vectors as ONE job (up to backend.max_circuits).
# Returns a list of CVaR costs, one per parameter vector.

def batch_evaluate_hw(
    params_list: list[np.ndarray],
    n: int,
    Np: int,
    Nq: int,
    N: int,
    L: int,
    shots: int,
    alpha: float,
    cost_type: str,
    sampler: Sampler,
    pm,
) -> list[float]:
    """
    Transpile and submit all circuits in params_list as a SINGLE IBM job.
    Returns CVaR costs in the same order as params_list.
    This collapses N queue submissions into 1.
    """
    # Build and transpile all circuits at once
    circuits_isa = [pm.run(build_linear_cnot_ansatz(N, L, p)) for p in params_list]

    # One job, many circuits
    job    = sampler.run([(qc, [], shots) for qc in circuits_isa])
    result = job.result()

    costs = []
    for i, pub_result in enumerate(result):
        counts = pub_result.data.meas.get_counts()
        eigenvalues = []
        for bs, count in counts.items():
            p_val, q_val = bitstring_to_pq(bs.replace(" ", ""), Np, Nq)
            eigenvalues.extend([cost_fn(p_val, q_val, n, cost_type)] * count)
        costs.append(cvar(eigenvalues, alpha))

    return costs
```

```python
# ── Simulate-to-convergence, validate on hardware (1 job per trial) ───────────

def run_vqe_sim_then_hw(
    n: int,
    backend,
    L: int = 3,
    shots: int = 2000,          # use more shots on hardware to beat noise
    alpha: float = 0.5,
    cost_type: str = "hamiltonian",
    max_iter: int = 50,
    n_validation_shots: int = 8192,
    verbose: bool = True,
) -> dict:
    """
    Phase 1 — Run full COBYLA VQE on local AerSimulator (fast, free).
    Phase 2 — Submit ONE job to IBM hardware with the optimised parameters.

    Queue wait: 1 job total per trial regardless of max_iter.
    """
    Np, Nq, N = get_qubit_counts(n)
    simulator  = AerSimulator()
    best_sim   = {"found": False, "p": None, "q": None}
    history_sim = []

    if verbose:
        print(f"\n{'─'*60}")
        print(f"  Phase 1 [SIM] n={n}, N={N} qubits, L={L}, α={alpha}")
        print(f"{'─'*60}")

    # ── Phase 1: simulate to convergence ─────────────────────────────────────
    params0 = np.random.uniform(-np.pi, np.pi, num_params(N, L))

    def sim_objective(params):
        cv,_ = evaluate_cost(
            params, n, Np, Nq, N, L, shots, alpha,
            cost_type, simulator, best_sim,
        )
        history_sim.append(cv)
        return cv

    optimizer = COBYLA(maxiter=max_iter)
    result    = optimizer.minimize(fun=sim_objective, x0=params0)
    opt_params = result.x

    if verbose:
        status = f"-> {best_sim['p']}×{best_sim['q']}" if best_sim["found"] else "X not found in sim"
        print(f"  Sim done — {status} | iters={len(history_sim)}")
        print(f"\n  Phase 2 [HW ] submitting 1 job to {backend.name} …")

    # ── Phase 2: single hardware validation job ───────────────────────────────
    pm      = generate_preset_pass_manager(optimization_level=3, backend=backend)
    sampler = Sampler(backend)

    qc_isa  = pm.run(build_linear_cnot_ansatz(N, L, opt_params))
    hw_job  = sampler.run([qc_isa], shots=n_validation_shots)
    hw_counts = hw_job.result()[0].data.meas.get_counts()

    hw_found = {"found": False, "p": None, "q": None}
    hw_eigenvalues = []
    for bs, count in hw_counts.items():
        p_val, q_val = bitstring_to_pq(bs.replace(" ", ""), Np, Nq)
        hw_eigenvalues.extend([cost_fn(p_val, q_val, n, cost_type)] * count)
        if p_val * q_val == n and p_val > 1 and q_val > 1:
            hw_found.update({"found": True, "p": p_val, "q": q_val})

    hw_cvar = cvar(hw_eigenvalues, alpha)

    if verbose:
        status_hw = f"-> {hw_found['p']}×{hw_found['q']}" if hw_found["found"] else "X not found on HW"
        print(f"  HW result — {status_hw} | CVaR={hw_cvar:.4f}")
        print(f"{'─'*60}\n")

    return {
        "found_sim":    best_sim["found"],
        "found_hw":     hw_found["found"],
        "p_sim":        best_sim["p"],   "q_sim": best_sim["q"],
        "p_hw":         hw_found["p"],   "q_hw":  hw_found["q"],
        "n":            n,
        "history_sim":  history_sim,     # full convergence curve (from sim)
        "hw_cvar":      hw_cvar,
        "hw_counts":    hw_counts,
        "backend_name": backend.name,
    }
```

```python
# ── Fig 4 — Hardware success rate, minimal queue usage ───────────────────────
import json

FIG4_INSTANCES = [
    # (n,   L,  alpha,  shots, max_iter, runs)
    #(15,    3,  0.50,   1000,  50,       5),
    #(21,    3,  0.50,   1000,  50,       5),
    #(57,    3,  0.50,   2000,  80,       5)
    (123,   3,  0.25,   1000,  100,      5),
    (253,   3,  0.25,   1000,  120,      3)
]

fig4_results = {}

for (n, L, alpha, shots, max_iter, runs) in FIG4_INSTANCES:
    _, _, N = get_qubit_counts(n)
    sim_successes, hw_successes = 0, 0
    print(f"\n{'═'*55}  n={n}  (N={N} qubits, {runs} runs)")

    for run_idx in range(runs):
        print(f"  Run {run_idx+1}/{runs} …")
        res = run_vqe_sim_then_hw(
            n, hw_backend, L=L, shots=shots,
            alpha=alpha, max_iter=max_iter, verbose=True,
        )
        if res["found_sim"]: sim_successes += 1
        if res["found_hw"]:  hw_successes  += 1

    fig4_results[n] = {
        "N":               N,
        "sim_success_rate": sim_successes / runs,
        "hw_success_rate":  hw_successes  / runs,
        "runs":            runs,
    }

with open("hardware_results1.json", "w") as f:
    json.dump(fig4_results, f, indent=2)
print("\nSaved hardware_results1.json")


with open("hardware_results.json", "r") as f:
    data_main = json.load(f)

with open("hardware_results1.json", "r") as f:
    data_new = json.load(f)

combined = {**data_main, **data_new}

fig4_results1 = combined

# ── Plot ──────────────────────────────────────────────────────────────────────
fig4, ax4 = plt.subplots(figsize=(7, 4.5))
n_vals = sorted(int(k) for k in fig4_results1.keys())
N_vals = [fig4_results1[str(n)]["N"] for n in n_vals]
sim_r  = [fig4_results1[str(n)]["sim_success_rate"] for n in n_vals]
hw_r   = [fig4_results1[str(n)]["hw_success_rate"]  for n in n_vals]

ax4.plot(N_vals, sim_r, color=COLORS[1], marker="s", ms=6, lw=2, ls="--", label="AerSimulator")
ax4.plot(N_vals, hw_r,  color=COLORS[0], marker="o", ms=7, lw=2,          label=f"IBM {hw_backend.name}")
for N_v, r in zip(N_vals, hw_r):
    ax4.annotate(f"{r:.0%}", (N_v, r), textcoords="offset points", xytext=(4, 5), fontsize=8)

ax4.set_xlabel("Number of Qubits N")
ax4.set_ylabel("Success Rate")
ax4.set_title("VQE Factorization Success Rate (COBYLA)")
ax4.set_ylim(-0.05, 1.10)
ax4.legend()
fig4.tight_layout()
fig4.savefig("hardware_success_rate.png", dpi=150, bbox_inches="tight")
plt.show()
print("hardware_success_rate.png")
```

```python
# Convergence on n=57

FIG5 = {"n": 253, "L": 3, "alpha": 0.25, "shots": 2000, "max_iter": 50}

# Run multiple trials so we can show mean ± std band (as in the paper)
N_TRIALS = 3
all_histories = []
hw_cvars, hw_found_flags = [], []

for trial in range(N_TRIALS):
    print(f"\nFig 5 trial {trial+1}/{N_TRIALS} …")
    res = run_vqe_sim_then_hw(
        FIG5["n"], hw_backend,
        L=FIG5["L"], shots=FIG5["shots"],
        alpha=FIG5["alpha"], max_iter=FIG5["max_iter"],
        n_validation_shots=8192, verbose=True,
    )
    all_histories.append(res["history_sim"])
    hw_cvars.append(res["hw_cvar"])
    hw_found_flags.append(res["found_hw"])

# Align histories to the same length (pad with last value)
max_len = max(len(h) for h in all_histories)
padded  = np.array([
    h + [h[-1]] * (max_len - len(h)) for h in all_histories
])
mean_curve = padded.mean(axis=0)
std_curve  = padded.std(axis=0)
iters      = np.arange(max_len)

# ── Plot ──────────────────────────────────────────────────────────────────────
fig5, ax5 = plt.subplots(figsize=(8, 4.5))

# Shaded mean ± std band (simulator convergence)
ax5.plot(iters, mean_curve, color=COLORS[1], lw=2, label="Simulator mean CVaR")
ax5.fill_between(iters,
                 mean_curve - std_curve,
                 mean_curve + std_curve,
                 color=COLORS[1], alpha=0.2, label="Simulator ±1σ")

# Hardware validation CVaR points at the last iteration
for trial_idx, (hw_cv, found) in enumerate(zip(hw_cvars, hw_found_flags)):
    marker = "★" if found else "✗"
    ax5.scatter(
        max_len - 1, hw_cv,
        color=COLORS[0], s=120, zorder=5,
        label=f"HW trial {trial_idx+1} {'(-> found)' if found else '(X not found)'}",
    )

ax5.set_xlabel("COBYLA Iteration")
ax5.set_ylabel(f"CVaR Cost  (α={FIG5['alpha']})")
ax5.set_title(
    f"Fig 5 — CVaR Convergence + Hardware Validation\n"
    f"n={FIG5['n']} (11×23), N={get_qubit_counts(FIG5['n'])[2]} qubits, "
    f"L={FIG5['L']}, shots={FIG5['shots']}"
)
ax5.set_yscale("symlog", linthresh=1)
ax5.legend(fontsize=8)
fig5.tight_layout()
fig5.savefig("hardware_convergence.png", dpi=150, bbox_inches="tight")
plt.show()
print("hardware_convergence.png")
```

```python
def warm_start_vqe(n, L=4, shots=10000, alpha=0.01, max_iter=300):
    """
    Warm-start strategy for inverse cost function:
    Stage 1 - Optimize with Hamiltonian cost (easy landscape, finds basin)
    Stage 2 - Use those parameters to initialize inverse cost (sharp gap, fast convergence)
    """
    Np, Nq, N = get_qubit_counts(n)
    
    print(f"n={n}, qubits={N}, Np={Np}, Nq={Nq}")
    
    # ── Stage 1: Hamiltonian warm-up ──────────────────────────────────────
    print("\nStage 1: Running Hamiltonian VQE to get warm parameters...")
    res_warm = run_vqe(n, optimizer="COBYLA",
                       L=L, shots=shots, alpha=alpha,
                       cost_type="hamiltonian",
                       max_iter=max_iter//3,      # short run, just find the basin
                       verbose=False)
    
    warm_params = res_warm["params"]              # extract optimized θ
    print(f"  Warm-start params obtained after {len(res_warm['history'])} iters")
    print(f"  Hamiltonian found solution: {res_warm['found']}")
    
    # ── Stage 2: Inverse cost from warm params ────────────────────────────
    print("\nStage 2: Running Inverse VQE from warm parameters...")
    res_inverse_warm = run_vqe(n, optimizer="NFT",
                                L=L, shots=shots, alpha=alpha,
                                cost_type="inverse",
                                max_iter=max_iter,
                                initial_params=warm_params,  # <── key change
                                verbose=False)
    
    # ── Baseline: Inverse cost from random params (paper's approach) ──────
    print("\nBaseline: Running Inverse VQE from random parameters...")
    res_inverse_rand = run_vqe(n, optimizer="NFT",
                                L=L, shots=shots, alpha=alpha,
                                cost_type="inverse",
                                max_iter=max_iter,
                                initial_params=None,          # random init
                                verbose=False)
    
    return res_warm, res_inverse_warm, res_inverse_rand


def plot_warm_start_comparison(n, L=4, shots=10000, alpha=0.01, max_iter=300):
    
    res_warm, res_inv_warm, res_inv_rand = warm_start_vqe(
        n, L=L, shots=shots, alpha=alpha, max_iter=max_iter
    )
    
    fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
    
    # ── Plot 1: Hamiltonian warm-up trajectory ────────────────────────────
    ax = axes[0]
    ax.plot(res_warm["history"], color=COLORS[0], lw=2)
    if res_warm["found"]:
        for i, v in enumerate(res_warm["history"]):
            if v < 1.0:
                ax.axvline(i, color=COLORS[0], ls=":", lw=1.5,
                           label=f"Solution @ iter {i}")
                break
    ax.set_title("Stage 1: Hamiltonian Warm-up")
    ax.set_xlabel("Iterations")
    ax.set_ylabel("CVaR Cost")
    ax.set_yscale("symlog")
    ax.legend()
    
    # ── Plot 2: Inverse — warm start vs random ────────────────────────────
    ax = axes[1]
    ax.plot(res_inv_warm["history"], color=COLORS[0], lw=2,
            label="Warm start (from Hamiltonian)")
    ax.plot(res_inv_rand["history"], color=COLORS[1], lw=2,
            ls="--", label="Random init (baseline)")
    
    # mark convergence points
    for res, color in [(res_inv_warm, COLORS[0]), (res_inv_rand, COLORS[1])]:
        if res["found"]:
            for i, v in enumerate(res["history"]):
                if v < -900:
                    ax.axvline(i, color=color, ls=":", lw=1.5)
                    break
    
    ax.set_title("Stage 2: Inverse Cost — Warm vs Random Init")
    ax.set_xlabel("Iterations")
    ax.set_ylabel("CVaR Cost")
    ax.set_yscale("symlog")
    ax.legend()
    
    # ── Plot 3: Summary bar chart ─────────────────────────────────────────
    ax = axes[2]
    
    def first_convergence(hist, threshold=-900):
        for i, v in enumerate(hist):
            if v < threshold:
                return i
        return max_iter  # didn't converge
    
    configs = ["Hamiltonian\n(warm-up)", "Inverse\n(warm start)", "Inverse\n(random)"]
    iters   = [
        first_convergence(res_warm["history"],    threshold=1.0),
        first_convergence(res_inv_warm["history"], threshold=-900),
        first_convergence(res_inv_rand["history"], threshold=-900),
    ]
    colors  = [COLORS[0], COLORS[0], COLORS[1]]
    bars    = ax.bar(configs, iters, color=colors, alpha=0.8, edgecolor="white")
    
    for bar, val in zip(bars, iters):
        label = str(val) if val < max_iter else "No conv."
        ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 2,
                label, ha="center", va="bottom", fontsize=10, color="white")
    
    ax.axhline(max_iter, color="gray", ls="--", lw=1, label="max_iter ceiling")
    ax.set_ylabel("Iterations to Convergence")
    ax.set_title("Convergence Speed Comparison")
    ax.legend()
    
    fig.suptitle(
        f"Warm-Start VQE for Inverse Cost Function  (n={n}, α={alpha}, shots={shots})",
        fontsize=13
    )
    fig.tight_layout()
    
    filename = f"warm_start_comparison_n{n}.png"
    fig.savefig(filename, dpi=150, bbox_inches="tight")
    print(f"\nSaved: {filename}")
    
    # ── Print summary ─────────────────────────────────────────────────────
    print(f"\n{'='*50}")
    print(f"RESULTS SUMMARY  (n={n})")
    print(f"{'='*50}")
    print(f"Hamiltonian warm-up converged:      {res_warm['found']}")
    print(f"Inverse (warm start) converged:     {res_inv_warm['found']}")
    print(f"Inverse (random init) converged:    {res_inv_rand['found']}")
    print(f"Iters — Hamiltonian warm-up:        {iters[0]}")
    print(f"Iters — Inverse warm start:         {iters[1]}")
    print(f"Iters — Inverse random:             {iters[2]}")
    
    speedup = iters[2] / iters[1] if iters[1] > 0 else float("inf")
    print(f"Speedup from warm start:            {speedup:.1f}x")
    
    return res_warm, res_inv_warm, res_inv_rand


# ── Run it ────────────────────────────────────────────────────────────────
res_warm, res_inv_warm, res_inv_rand = plot_warm_start_comparison(
    n=511, L=4, shots=10000, alpha=0.01, max_iter=300
)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def total_evals(res_warm, res_inv):
    return len(res_warm["history"]) + len(res_inv["history"])


def first_convergence(hist, threshold, max_iter):
    for i, v in enumerate(hist):
        if v < threshold:
            return i
    return max_iter


def run_experiment_multi_seed(n, L=4, shots=10000, alpha=0.01,
                              max_iter=300, n_trials=20):

    warm_totals = []
    rand_totals = []

    warm_success = 0
    rand_success = 0

    for seed in range(n_trials):
        print(f"\n=== Trial {seed+1}/{n_trials} ===")

        # ── Stage 1: Hamiltonian warm-up ─────────────────────
        res_warm = run_vqe(
            n, L=L, shots=shots, alpha=alpha,
            cost_type="hamiltonian",
            max_iter=max_iter // 3,
            optimizer="COBYLA",
            #seed=seed,
            verbose=False
        )

        warm_params = res_warm["params"]

        # ── Stage 2: inverse (warm start) ────────────────────
        res_inv_warm = run_vqe(
            n, L=L, shots=shots, alpha=alpha,
            cost_type="inverse",
            max_iter=max_iter,
            optimizer="NFT",
            initial_params=warm_params,
            #seed=seed,
            verbose=False
        )

        # ── Baseline: inverse (random) ───────────────────────
        res_inv_rand = run_vqe(
            n, L=L, shots=shots, alpha=alpha,
            cost_type="inverse",
            max_iter=max_iter,
            optimizer="NFT",
            initial_params=None,
            #seed=seed,
            verbose=False
        )

        # ── Compute total cost-to-solution ───────────────────
        thresh_inv = -900

        warm_conv = first_convergence(res_inv_warm["history"], thresh_inv, max_iter)
        rand_conv = first_convergence(res_inv_rand["history"], thresh_inv, max_iter)

        total_warm = len(res_warm["history"]) + warm_conv
        total_rand = rand_conv

        warm_totals.append(total_warm)
        rand_totals.append(total_rand)

        warm_success += int(res_inv_warm["found"])
        rand_success += int(res_inv_rand["found"])

    # ── Convert to arrays ───────────────────────────────────
    warm_totals = np.array(warm_totals)
    rand_totals = np.array(rand_totals)

    # ── Summary stats ───────────────────────────────────────
    print("\n" + "="*60)
    print("MULTI-SEED RESULTS")
    print("="*60)

    print(f"Warm start success rate: {warm_success}/{n_trials} = {warm_success/n_trials:.2f}")
    print(f"Random init success rate: {rand_success}/{n_trials} = {rand_success/n_trials:.2f}")

    print("\nTotal evaluations to convergence:")
    print(f"Warm start: mean = {warm_totals.mean():.1f}, std = {warm_totals.std():.1f}")
    print(f"Random init: mean = {rand_totals.mean():.1f}, std = {rand_totals.std():.1f}")

    speedup = rand_totals.mean() / warm_totals.mean()
    print(f"\nAverage speedup (total cost): {speedup:.2f}x")

    return warm_totals, rand_totals


def plot_multi_seed_results(warm_totals, rand_totals, n):

    fig, ax = plt.subplots(figsize=(6, 4.5))

    data = [warm_totals, rand_totals]

    ax.boxplot(data, tick_labels=["Warm start", "Random init"])

    ax.set_ylabel("Total evaluations to convergence")
    ax.set_title(f"Total Cost-to-Solution (n={n})")

    plt.tight_layout()
    filename = f"multi_seed_total_cost_n{n}.png"
    plt.savefig(filename, dpi=150)
    print(f"Saved: {filename}")

warm_totals, rand_totals = run_experiment_multi_seed(
    n=57,
    L=3,
    shots=1000,
    alpha=0.01,
    max_iter=100000,
    n_trials=20
)

plot_multi_seed_results(warm_totals, rand_totals, n=57)
```

```python
"""
Warm-Start Strategies for Inverse Cost VQE — Fixed Implementation
=================================================================
Based on the open problem from:
  Sobhani et al. PRA 111, 042413 (2025)

IMPORTANT: n=15 and n=21 are TOO EASY for this comparison.
With only 16/32 computational basis states, random initialisation
finds the solution trivially (~100% rate, 0 iterations needed).
The warm-start advantage only becomes visible at n >= 57 (N=7 qubits)
where the solution bitstring is 1 of 128 states (~1.6% base rate).

Three strategies implemented and fairly compared:
  Baseline   — pure random init, inverse cost, full budget
  Strategy 1 — Hamiltonian pre-train → switch to inverse (cost-switching)
  Strategy 2 — Structure-inspired init → inverse cost
  Strategy 3 — Mean-field init → inverse cost
  Strategy 4 — Multi-start screening → best params → inverse cost

Metrics tracked per run:
  - found:            did the correct (p,q) ever appear in any shot?
  - first_found_iter: iteration index when solution FIRST appeared
  - final_fidelity:   fraction of last-iteration shots giving correct solution
  - cvar_history:     full CVaR trajectory for plotting

Usage:
  python vqe_warmstart.py              # default n=57
  python vqe_warmstart.py --n 123     # harder instance
"""

import sys
import argparse
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from math import ceil, log2, asin, sqrt
from itertools import product as iproduct
from scipy.optimize import minimize
import warnings
warnings.filterwarnings("ignore")

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# ── Config ─────────────────────────────────────────────────────────────────────
SIM    = AerSimulator()
COLORS = {
    "baseline":  "#d62728",
    "s1_ham":    "#1f77b4",
    "s1_inv":    "#ff7f0e",
    "s2":        "#2ca02c",
    "s3":        "#9467bd",
    "s4":        "#8c564b",
}
COLOR_LIST = ["#d62728", "#ff7f0e", "#2ca02c", "#9467bd", "#8c564b"]
plt.rcParams.update({
    "font.family": "DejaVu Sans", "axes.spines.top": False,
    "axes.spines.right": False, "axes.grid": True,
    "grid.alpha": 0.3, "axes.labelsize": 11, "axes.titlesize": 12,
})


# ═══════════════════════════════════════════════════════════════════════════════
# Problem encoding
# ═══════════════════════════════════════════════════════════════════════════════

def get_qubit_counts(n):
    B = ceil(log2(n + 1)); Np = ceil(log2(n**0.5 + 1)); Nq = B - 1
    return Np, Nq, (Np - 1) + (Nq - 1)

def bitstring_to_pq(bits, Np, Nq):
    bits = bits[::-1]
    q_bits = [int(b) for b in bits[:Nq - 1]]
    p_bits = [int(b) for b in bits[Nq - 1:]]
    p = 1 + sum(p_bits[j] * (2**(j+1)) for j in range(len(p_bits)))
    q = 1 + sum(q_bits[k] * (2**(k+1)) for k in range(len(q_bits)))
    return p, q

def cost_fn(p, q, n, ct):
    d = abs(n - p * q)
    if ct == "hamiltonian": return float(d ** 2)
    if ct == "inverse":     return -1.0 / (d + 0.001)
    raise ValueError(ct)

def build_ansatz(N, L, params):
    qc = QuantumCircuit(N); qc.h(range(N)); idx = 0
    for layer in range(L):
        for q in range(N): qc.ry(float(params[idx]), q); idx += 1
        if layer < L - 1:
            for q in range(N - 1): qc.cx(q, q + 1)
    qc.measure_all()
    return qc

def cvar(evals, alpha):
    s = sorted(evals); k = max(1, int(ceil(alpha * len(s))))
    return float(np.mean(s[:k]))

def get_true_factors(n):
    for p in range(2, int(n**0.5) + 1):
        if n % p == 0:
            return p, n // p
    raise ValueError(f"{n} is not semiprime")


# ═══════════════════════════════════════════════════════════════════════════════
# Core circuit runner with proper iteration tracking
# ═══════════════════════════════════════════════════════════════════════════════

def run_circuit_detailed(params, n, Np, Nq, N, L, shots, alpha, ct):
    counts = SIM.run(build_ansatz(N, L, params), shots=shots).result().get_counts()
    p_true, q_true = get_true_factors(n)
    evals = []; correct_shots = 0
    for bs, cnt in counts.items():
        p, q = bitstring_to_pq(bs, Np, Nq)
        evals.extend([cost_fn(p, q, n, ct)] * cnt)
        if (p == p_true and q == q_true) or (p == q_true and q == p_true):
            correct_shots += cnt
    fidelity = correct_shots / shots
    return cvar(evals, alpha), correct_shots > 0, fidelity


def run_vqe(n, params0, L=3, shots=1000, alpha=0.25, ct="inverse",
            max_iter=150, label=""):
    """
    Run VQE. Tracks:
      - cvar_history[i]     : CVaR at evaluation i (0-indexed)
      - fidelity_history[i] : P(correct) at evaluation i
      - first_found_iter    : eval index where solution FIRST appeared
                              None if never found within budget
    """
    Np, Nq, N = get_qubit_counts(n)
    cvar_hist     = []
    fidelity_hist = []
    first_found   = [None]

    def obj(params):
        cv, found, fid = run_circuit_detailed(
            params, n, Np, Nq, N, L, shots, alpha, ct)
        # Record BEFORE appending so index = current call count
        if found and first_found[0] is None:
            first_found[0] = len(cvar_hist)   # 0-indexed eval where found
        cvar_hist.append(cv)
        fidelity_hist.append(fid)
        return cv

    minimize(obj, params0, method="COBYLA",
             options={"maxiter": max_iter, "rhobeg": 0.5})

    found     = first_found[0] is not None
    final_fid = fidelity_hist[-1] if fidelity_hist else 0.0

    if label:
        status = (f"found @ eval {first_found[0]}"
                  if found else f"NOT found in {len(cvar_hist)} evals")
        print(f"    [{label:32s}]  {status:<30s}  fidelity={final_fid:.4f}")

    return {
        "found":            found,
        "first_found_iter": first_found[0],
        "total_evals":      len(cvar_hist),
        "final_fidelity":   final_fid,
        "cvar_history":     cvar_hist,
        "fidelity_history": fidelity_hist,
    }


# ═══════════════════════════════════════════════════════════════════════════════
# Initialisation strategies
# ═══════════════════════════════════════════════════════════════════════════════

def random_init(N, L):
    return np.random.uniform(-np.pi, np.pi, N * L)


def hamiltonian_pretrain_init(n, L, shots, alpha, pretrain_iters):
    """
    Strategy 1: short Hamiltonian VQE → transfer params to inverse cost.
    Returns: (warm_params, ham_cvar_history)
    """
    Np, Nq, N = get_qubit_counts(n)
    params0 = random_init(N, L)
    ham_hist = []

    def obj_ham(p):
        cv, _, _ = run_circuit_detailed(p, n, Np, Nq, N, L, shots, alpha, "hamiltonian")
        ham_hist.append(cv)
        return cv

    result = minimize(obj_ham, params0, method="COBYLA",
                      options={"maxiter": pretrain_iters, "rhobeg": 0.5})
    return result.x, ham_hist


def structure_inspired_init(n, L):
    """
    Strategy 2: analytically bias layer-1 RY angles toward p ≈ √n.
    p-register → encode round(√n) in binary.
    q-register → uniform superposition (π/2).
    Deeper layers → small random perturbations.
    """
    Np, Nq, N = get_qubit_counts(n)
    p_est      = int(round(sqrt(n)))
    n_p_qubits = Np - 1
    n_q_qubits = Nq - 1
    p_bits = [(p_est >> j) & 1 for j in range(1, Np)]
    p_bits = (p_bits + [0] * n_p_qubits)[:n_p_qubits]

    layer1 = np.zeros(N)
    for j, b in enumerate(p_bits):
        layer1[j] = np.pi if b else 0.0
    for k in range(n_q_qubits):
        layer1[n_p_qubits + k] = np.pi / 2

    params = np.zeros(N * L)
    params[:N] = layer1
    rng = np.random.default_rng()
    for layer in range(1, L):
        params[layer * N:(layer + 1) * N] = rng.uniform(-0.25, 0.25, N)
    return params


def mean_field_init(n, L, n_restarts=8):
    """
    Strategy 3: classical mean-field optimisation over product states.
    Returns: (params, marginal_probs, expected_cost)
    """
    Np, Nq, N = get_qubit_counts(n)
    cost_table = {}
    for bits in iproduct([0, 1], repeat=N):
        bs = "".join(str(b) for b in bits)
        p, q = bitstring_to_pq(bs, Np, Nq)
        cost_table[bits] = cost_fn(p, q, n, "hamiltonian")

    def expected_cost(probs):
        total = 0.0
        pc = np.clip(probs, 1e-8, 1 - 1e-8)
        for bits, c in cost_table.items():
            prob = np.prod([pc[i] if b else (1 - pc[i])
                            for i, b in enumerate(bits)])
            total += prob * c
        return total

    best_probs = None; best_val = np.inf
    for _ in range(n_restarts):
        p0  = np.random.uniform(0.1, 0.9, N)
        res = minimize(expected_cost, p0, method="L-BFGS-B",
                       bounds=[(0.01, 0.99)] * N,
                       options={"maxiter": 300, "ftol": 1e-10})
        if res.fun < best_val:
            best_val = res.fun; best_probs = res.x

    layer1 = np.array([2 * asin(sqrt(np.clip(pi, 0, 1))) for pi in best_probs])
    params = np.zeros(N * L)
    params[:N] = layer1
    rng = np.random.default_rng()
    for layer in range(1, L):
        params[layer * N:(layer + 1) * N] = rng.uniform(-0.2, 0.2, N)
    return params, best_probs, best_val


def multistart_screen_init(n, L, shots, alpha, K=8, screen_iters=20):
    """
    Strategy 4: screen K random starts with Hamiltonian cost, pick best.
    Returns: (best_params, all_candidate_scores)
    """
    Np, Nq, N = get_qubit_counts(n)
    candidates = []
    for _ in range(K):
        p0 = random_init(N, L)
        hist = []

        def obj_s(p, _h=hist):
            cv, _, _ = run_circuit_detailed(p, n, Np, Nq, N, L, shots, alpha, "hamiltonian")
            _h.append(cv); return cv

        res = minimize(obj_s, p0, method="COBYLA",
                       options={"maxiter": screen_iters, "rhobeg": 0.5})
        candidates.append((res.fun, res.x))

    candidates.sort(key=lambda x: x[0])
    return candidates[0][1], [v for v, _ in candidates]


# ═══════════════════════════════════════════════════════════════════════════════
# Main comparison
# ═══════════════════════════════════════════════════════════════════════════════

def run_comparison(n, L=3, shots=1000, alpha=0.25, max_iter=150,
                   pretrain_iters=40, screen_K=8, screen_iters=20, runs=5):

    Np, Nq, N = get_qubit_counts(n)
    p_true, q_true = get_true_factors(n)

    print(f"\n{'='*62}")
    print(f"  Warm-Start Comparison  (n={n} = {p_true}×{q_true})")
    print(f"  N={N} qubits  |  2^N={2**N} states  |  "
          f"Solution prob (random) ≈ {2/2**N:.2%}")
    print(f"  L={L}  shots={shots}  α={alpha}  max_iter={max_iter}")
    print(f"{'='*62}")

    STRATEGY_KEYS = ["Baseline\n(random)", "S1: Ham\npre-train",
                     "S2: Structure\ninit", "S3: Mean-field\ninit",
                     "S4: Multi-start\nscreen"]

    all_results      = {k: [] for k in STRATEGY_KEYS}
    pretrain_hists   = []
    mf_probs_list    = []

    for run_i in range(runs):
        print(f"\n  ── Run {run_i+1}/{runs} ──")

        # Baseline
        r = run_vqe(n, random_init(N, L), L=L, shots=shots, alpha=alpha,
                    ct="inverse", max_iter=max_iter, label="Baseline (random)")
        all_results["Baseline\n(random)"].append(r)

        # S1: Hamiltonian pre-train
        warm_p, ham_hist = hamiltonian_pretrain_init(n, L, shots, alpha, pretrain_iters)
        pretrain_hists.append(ham_hist)
        r1 = run_vqe(n, warm_p, L=L, shots=shots, alpha=alpha, ct="inverse",
                     max_iter=max_iter, label="S1: Ham pre-train → inverse")
        # Account for pretrain cost in total-evals metric
        r1["pretrain_evals"] = len(ham_hist)
        if r1["first_found_iter"] is not None:
            r1["first_found_iter_total"] = r1["first_found_iter"] + len(ham_hist)
        else:
            r1["first_found_iter_total"] = None
        all_results["S1: Ham\npre-train"].append(r1)

        # S2: Structure-inspired
        r2 = run_vqe(n, structure_inspired_init(n, L), L=L, shots=shots,
                     alpha=alpha, ct="inverse", max_iter=max_iter,
                     label="S2: Structure-inspired init")
        all_results["S2: Structure\ninit"].append(r2)

        # S3: Mean-field
        mf_p, mf_probs, mf_val = mean_field_init(n, L)
        mf_probs_list.append(mf_probs)
        r3 = run_vqe(n, mf_p, L=L, shots=shots, alpha=alpha, ct="inverse",
                     max_iter=max_iter, label="S3: Mean-field init")
        r3["mf_probs"] = mf_probs; r3["mf_val"] = mf_val
        all_results["S3: Mean-field\ninit"].append(r3)

        # S4: Multi-start screen
        sc_p, sc_scores = multistart_screen_init(n, L, shots, alpha,
                                                  K=screen_K, screen_iters=screen_iters)
        r4 = run_vqe(n, sc_p, L=L, shots=shots, alpha=alpha, ct="inverse",
                     max_iter=max_iter, label="S4: Multi-start screen")
        total_screen_evals = screen_K * screen_iters
        if r4["first_found_iter"] is not None:
            r4["first_found_iter_total"] = r4["first_found_iter"] + total_screen_evals
        else:
            r4["first_found_iter_total"] = None
        r4["screen_scores"] = sc_scores
        all_results["S4: Multi-start\nscreen"].append(r4)

    return all_results, STRATEGY_KEYS, pretrain_hists, mf_probs_list


# ═══════════════════════════════════════════════════════════════════════════════
# Summary
# ═══════════════════════════════════════════════════════════════════════════════

def print_summary(all_results, strategy_keys, n):
    p_true, q_true = get_true_factors(n)
    print(f"\n{'='*68}")
    print(f"  RESULTS SUMMARY  (n={n} = {p_true}×{q_true})")
    print(f"{'='*68}")
    print(f"  {'Strategy':30s}  {'SR':>5}  {'1st-found (total)':>18}  {'Fidelity':>9}")
    print(f"  {'-'*64}")

    baseline_iters = []
    for r in all_results["Baseline\n(random)"]:
        if r["found"] and r["first_found_iter"] is not None:
            baseline_iters.append(r["first_found_iter"])
    baseline_avg = np.mean(baseline_iters) if baseline_iters else float("nan")

    for key in strategy_keys:
        runs_data = all_results[key]
        n_runs    = len(runs_data)
        successes = sum(1 for r in runs_data if r["found"])
        sr        = successes / n_runs

        found_its = []
        for r in runs_data:
            if r["found"]:
                it = r.get("first_found_iter_total", r["first_found_iter"])
                if it is not None: found_its.append(it)

        avg_iter = np.mean(found_its) if found_its else float("nan")
        avg_fid  = np.mean([r["final_fidelity"] for r in runs_data])
        label    = key.replace("\n", " ")

        speedup  = ""
        if key != "Baseline\n(random)" and not np.isnan(avg_iter) and not np.isnan(baseline_avg):
            if avg_iter > 0:
                speedup = f"  ({baseline_avg/avg_iter:.1f}× faster)" if avg_iter < baseline_avg else ""

        print(f"  {label:30s}  {sr:>5.0%}  {avg_iter:>18.1f}{speedup}")
        print(f"  {'':30s}  {'':>5}  {'':>18}  {avg_fid:>9.4f}")

    print(f"{'='*68}\n")


# ═══════════════════════════════════════════════════════════════════════════════
# Plots
# ═══════════════════════════════════════════════════════════════════════════════

def plot_all(all_results, strategy_keys, n, pretrain_hists, mf_probs_list,
             tag=""):
    p_true, q_true = get_true_factors(n)
    Np, Nq, N = get_qubit_counts(n)

    # ── Figure 1: Main comparison ─────────────────────────────────────────────
    fig = plt.figure(figsize=(18, 11))
    fig.suptitle(
        f"Warm-Start Strategies for Inverse Cost VQE\n"
        f"n={n} ({p_true}×{q_true}),  N={N} qubits,  α=0.25,  "
        f"Inverse cost: −1/(|n−pq|+ε)",
        fontsize=13,
    )
    gs = fig.add_gridspec(2, 3, hspace=0.42, wspace=0.35)

    # Panel 1: Success rate
    ax1 = fig.add_subplot(gs[0, 0])
    srs = [sum(r["found"] for r in all_results[k]) / len(all_results[k])
           for k in strategy_keys]
    short = [k.replace("\n", "\n") for k in strategy_keys]
    bars  = ax1.bar(range(len(srs)), srs, color=COLOR_LIST,
                    edgecolor="white", alpha=0.88, width=0.65)
    ax1.set_xticks(range(len(srs))); ax1.set_xticklabels(short, fontsize=8)
    ax1.set_ylabel("Success Rate"); ax1.set_ylim(0, 1.25)
    ax1.set_title("Success Rate  (inverse cost)")
    for i, sr in enumerate(srs):
        ax1.text(i, sr + 0.04, f"{sr:.0%}", ha="center",
                 fontsize=10, fontweight="bold")

    # Panel 2: Total evals to first solution
    ax2 = fig.add_subplot(gs[0, 1])
    avg_iters = []
    for k in strategy_keys:
        its = []
        for r in all_results[k]:
            if r["found"]:
                it = r.get("first_found_iter_total", r["first_found_iter"])
                if it is not None: its.append(it)
        avg_iters.append(np.mean(its) if its else np.nan)

    for i, (v, c) in enumerate(zip(avg_iters, COLOR_LIST)):
        if not np.isnan(v):
            ax2.bar(i, v, color=c, edgecolor="white", alpha=0.88, width=0.65)
            ax2.text(i, v + 0.5, f"{v:.0f}", ha="center",
                     fontsize=9, fontweight="bold")
    ax2.set_xticks(range(len(strategy_keys)))
    ax2.set_xticklabels(short, fontsize=8)
    ax2.set_ylabel("Total Function Evals (incl. overhead)")
    ax2.set_title("Evals to First Solution\n(incl. pre-train/screen cost)")

    # Panel 3: Final fidelity box plot
    ax3 = fig.add_subplot(gs[0, 2])
    fid_data = [[r["final_fidelity"] for r in all_results[k]]
                for k in strategy_keys]
    bp = ax3.boxplot(fid_data, patch_artist=True,
                     medianprops={"linewidth": 2, "color": "white"})
    for patch, c in zip(bp["boxes"], COLOR_LIST):
        patch.set_facecolor(c); patch.set_alpha(0.75)
    ax3.set_xticks(range(1, len(strategy_keys) + 1))
    ax3.set_xticklabels(short, fontsize=8)
    ax3.set_ylabel("P(correct bitstring) in final circuit")
    ax3.set_title("Final Fidelity Distribution")

    # Panel 4: CVaR trajectories
    ax4 = fig.add_subplot(gs[1, :2])
    for k, c in zip(strategy_keys, COLOR_LIST):
        hists = [r["cvar_history"] for r in all_results[k]]
        if not hists: continue
        max_len = max(len(h) for h in hists)
        padded  = np.array([h + [h[-1]] * (max_len - len(h)) for h in hists])
        mean_h  = np.mean(padded, axis=0)
        std_h   = np.std(padded, axis=0)
        xs      = np.arange(len(mean_h))
        label   = k.replace("\n", " ")
        ax4.plot(xs, mean_h, color=c, lw=2, label=label)
        ax4.fill_between(xs, mean_h - std_h, mean_h + std_h, color=c, alpha=0.12)
    ax4.set_xlabel("Inverse-Cost Function Evaluations")
    ax4.set_ylabel("CVaR  (inverse cost, lower = better)")
    ax4.set_title("Mean CVaR Trajectory ± 1σ")
    ax4.set_yscale("symlog"); ax4.legend(fontsize=9, loc="upper right")

    # Panel 5: Fidelity trajectories
    ax5 = fig.add_subplot(gs[1, 2])
    for k, c in zip(strategy_keys, COLOR_LIST):
        hists = [r["fidelity_history"] for r in all_results[k]]
        if not hists: continue
        max_len = max(len(h) for h in hists)
        padded  = np.array([h + [h[-1]] * (max_len - len(h)) for h in hists])
        mean_h  = np.mean(padded, axis=0)
        ax5.plot(np.arange(len(mean_h)), mean_h, color=c, lw=2,
                 label=k.replace("\n", " "))
    ax5.set_xlabel("Inverse-Cost Function Evaluations")
    ax5.set_ylabel("P(correct factorisation per shot)")
    ax5.set_title("Solution Probability per Iteration")
    ax5.legend(fontsize=8)

    fname = f"warmstart_comparison_{tag}.png"
    fig.savefig(fname, dpi=150, bbox_inches="tight"); plt.close(fig)
    print(f"  Saved → {fname}")

    # ── Figure 2: S1 detail ───────────────────────────────────────────────────
    if pretrain_hists and all_results["S1: Ham\npre-train"]:
        fig2, axes2 = plt.subplots(1, 2, figsize=(13, 5))
        fig2.suptitle(f"Strategy 1: Hamiltonian Pre-train → Inverse  (n={n})",
                      fontsize=13)
        ax = axes2[0]
        ph1 = pretrain_hists[0]
        ph2 = all_results["S1: Ham\npre-train"][0]["cvar_history"]
        xs1 = np.arange(len(ph1))
        xs2 = np.arange(len(ph1), len(ph1) + len(ph2))
        ax.plot(xs1, ph1, color=COLORS["s1_ham"], lw=2,
                label="Phase 1: Hamiltonian CVaR")
        ax.axvline(len(ph1), color="grey", ls=":", lw=2,
                   label="Cost switch → inverse")
        ax.plot(xs2, ph2, color=COLORS["s1_inv"], lw=2,
                label="Phase 2: Inverse CVaR")
        ff = all_results["S1: Ham\npre-train"][0]["first_found_iter"]
        if ff is not None:
            ax.axvline(len(ph1) + ff, color=COLORS["s2"], ls="--", lw=1.5,
                       label=f"Solution found @ eval {len(ph1)+ff}")
        ax.set_xlabel("Cumulative Evals"); ax.set_ylabel("CVaR")
        ax.set_yscale("symlog"); ax.set_title("Example Run: Both Phases")
        ax.legend(fontsize=9)

        ax2 = axes2[1]
        for label_k, color_k in [("Baseline\n(random)", COLORS["baseline"]),
                                   ("S1: Ham\npre-train", COLORS["s1_inv"])]:
            hists = [r["cvar_history"] for r in all_results[label_k]]
            max_len = max(len(h) for h in hists)
            padded  = np.array([h + [h[-1]] * (max_len - len(h)) for h in hists])
            mean_h  = np.mean(padded, axis=0); std_h = np.std(padded, axis=0)
            xs      = np.arange(len(mean_h))
            ax2.plot(xs, mean_h, color=color_k, lw=2,
                     label=label_k.replace("\n", " "))
            ax2.fill_between(xs, mean_h - std_h, mean_h + std_h,
                             color=color_k, alpha=0.15)
        ax2.set_xlabel("Inverse-Cost Evals"); ax2.set_ylabel("CVaR")
        ax2.set_yscale("symlog"); ax2.set_title("Warm vs Cold: Mean ± 1σ")
        ax2.legend()
        fig2.tight_layout()
        f2 = f"warmstart_s1_detail_{tag}.png"
        fig2.savefig(f2, dpi=150, bbox_inches="tight"); plt.close(fig2)
        print(f"  Saved → {f2}")

    # ── Figure 3: Mean-field marginals ────────────────────────────────────────
    if mf_probs_list:
        mean_probs = np.mean(mf_probs_list, axis=0)
        fig3, axes3 = plt.subplots(1, 2, figsize=(12, 4.5))
        fig3.suptitle(f"Strategy 3: Mean-Field Initialisation  (n={n})", fontsize=13)
        ql = [f"q{i}" for i in range(N)]

        ax = axes3[0]
        bars = ax.bar(ql, mean_probs, color=COLORS["s3"],
                      edgecolor="white", alpha=0.85)
        ax.axhline(0.5, color="grey", ls="--", lw=1.2, label="Uniform (0.5)")
        ax.set_ylim(-0.05, 1.1); ax.set_xlabel("Qubit")
        ax.set_ylabel("Mean P(qubit=1)"); ax.set_title("Optimal Marginals")
        ax.legend()
        for bar, p in zip(bars, mean_probs):
            ax.text(bar.get_x() + bar.get_width()/2, p + 0.03,
                    f"{p:.2f}", ha="center", fontsize=9)

        theta = np.array([2 * asin(sqrt(np.clip(pi, 0, 1))) for pi in mean_probs])
        ax2   = axes3[1]
        ax2.bar(ql, theta, color=COLORS["s3"], edgecolor="white", alpha=0.85)
        ax2.axhline(np.pi/2, color="grey", ls="--", lw=1.2, label="π/2 (uniform)")
        ax2.set_xlabel("Qubit"); ax2.set_ylabel("θ = 2·arcsin(√p)  (rad)")
        ax2.set_title("Layer-1 RY Angles from Mean-Field"); ax2.legend()
        fig3.tight_layout()
        f3 = f"warmstart_s3_marginals_{tag}.png"
        fig3.savefig(f3, dpi=150, bbox_inches="tight"); plt.close(fig3)
        print(f"  Saved → {f3}")


# ═══════════════════════════════════════════════════════════════════════════════
# Main
# ═══════════════════════════════════════════════════════════════════════════════
if __name__ == "__main__":
    # Hardcode params instead of argparse
    class Args:
        n      = 4087   # 61×67, N=15 qubits, 1024 states
        runs   = 5
        shots  = 150
        alpha  = 0.1
        iters  = 1000
        L      = 4
    args = Args()

    _, __, N = get_qubit_counts(args.n)
    
    all_results, keys, pretrain_hists, mf_probs = run_comparison(
        n              = args.n,
        L              = args.L,
        shots          = args.shots,
        alpha          = args.alpha,
        max_iter       = args.iters,
        pretrain_iters = 40,
        screen_K       = 8,
        screen_iters   = 20,
        runs           = args.runs,
    )

    print_summary(all_results, keys, args.n)
    plot_all(all_results, keys, args.n, pretrain_hists, mf_probs,
             tag=f"n{args.n}")
```
